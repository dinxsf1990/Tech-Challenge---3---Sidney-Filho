###*Resolver incompatibilidades técnicas*

```python
def converter_colunas_para_string(df):
    return df.select([
        F.col(f"`{coluna.replace('`', '``')}`")
         .cast("string")
         .alias(coluna)
        for coluna in df.columns
    ])

df2023_str = converter_colunas_para_string(df2023)
df2024_str = converter_colunas_para_string(df2024)
df2025_str = converter_colunas_para_string(df2025)
```

###*Consolidar*

```python
df_teste = df2023_str.unionByName(
    df2024_str,
    allowMissingColumns=True
)

df_total = df_teste.unionByName(
    df2025_str,
    allowMissingColumns=True
)
```

## Silver

###*Mapeamento dos Campos*

As pesquisas possuem nomenclaturas diferentes entre os anos.

Exemplos:

2023:
('P1_a ', 'Idade')

2024/2025:
1.a_idade

A camada Silver unifica esses campos utilizando coalesce().

###*Consolidar os campos*

```python
silver = df_total.select(

    F.col("ano_pesquisa"),

    F.coalesce(
        F.col("`('P1_a ', 'Idade')`"),
        F.col("`1.a_idade`")
    ).alias("idade"),

    F.coalesce(
        F.col("`('P1_b ', 'Genero')`"),
        F.col("`1.b_genero`")
    ).alias("genero"),

    F.coalesce(
        F.col("`('P1_i_2 ', 'Regiao onde mora')`"),
        F.col("`1.i.2_regiao_onde_mora`")
    ).alias("regiao"),

    F.coalesce(
        F.col("`('P2_f ', 'Cargo Atual')`"),
        F.col("`2.f_cargo_atual`")
    ).alias("cargo"),

    F.coalesce(
        F.col("`('P2_r ', 'Atualmente qual a sua forma de trabalho?')`"),
        F.col("`2.q_modelo_de_trabalho_atual`"),
        F.col("`2.r_modelo_de_trabalho_atual`")
    ).alias("modelo_trabalho"),

    F.coalesce(
        F.col("`4.m_usa_chatgpt_ou_copilot_no_trabalho?`"),
        F.col("`4.j_usa_chatgpt_ou_copilot_no_trabalho?`")
    ).alias("uso_ia")

)
```

###**Adoção de Inteligência Artificial**

### Pergunta de Negócio

Como os profissionais da área de Dados utilizam Inteligência Artificial no dia a dia?

### Objetivo

Avaliar os diferentes modelos de utilização de IA generativa.

### Metodologia

As respostas foram agrupadas em:

- IA Gratuita
- IA Corporativa
- IA para Código
- IA Paga Individual
- Não utiliza IA

```python
gold_ia = (
    silver
    .where(F.col("uso_ia").isNotNull())
    .groupBy("uso_ia")
    .count()
    .orderBy(F.desc("count"))
)

gold_ia.show(
    truncate=False
)
```

```python
gold_ia_pd = gold_ia.toPandas()

def classificar_ia(texto):

    texto = str(texto).lower()

    if "não utilizo" in texto:
        return "Não utiliza IA"

    elif "copilot" in texto or "cursor" in texto:
        return "Copilot / IA para Código"

    elif "empresa em que trabalho paga" in texto:
        return "IA Corporativa"

    elif "pago do meu próprio bolso" in texto:
        return "IA Paga Individual"

    elif "gratuitas" in texto:
        return "IA Gratuita"

    return "Outros"

gold_ia_pd["categoria"] = (
    gold_ia_pd["uso_ia"]
    .apply(classificar_ia)
)

gold_ia_exec = (
    gold_ia_pd
    .groupby("categoria")["count"]
    .sum()
    .reset_index()
    .sort_values(
        by="count",
        ascending=False
    )
)

gold_ia_exec
```

```python
silver_ia = (
    silver

    .withColumn(
        "ia_nao_utiliza",
        F.when(
            F.lower(F.col("uso_ia")).contains("não utilizo"),
            1
        ).otherwise(0)
    )

    .withColumn(
        "ia_gratuita",
        F.when(
            F.lower(F.col("uso_ia")).contains("gratuit"),
            1
        ).otherwise(0)
    )

    .withColumn(
        "ia_paga_individual",
        F.when(
            F.lower(F.col("uso_ia")).contains("próprio bolso"),
            1
        ).otherwise(0)
    )

    .withColumn(
        "ia_corporativa",
        F.when(
            F.lower(F.col("uso_ia")).contains("empresa em que trabalho paga"),
            1
        ).otherwise(0)
    )

    .withColumn(
        "ia_codigo",
        F.when(
            F.lower(F.col("uso_ia")).contains("copilot")
            |
            F.lower(F.col("uso_ia")).contains("cursor")
            |
            F.lower(F.col("uso_ia")).contains("claude code"),
            1
        ).otherwise(0)
    )
)
```

```python
silver_ia.agg(

    F.sum("ia_nao_utiliza").alias("nao_utiliza"),

    F.sum("ia_gratuita").alias("gratuita"),

    F.sum("ia_paga_individual").alias("paga_individual"),

    F.sum("ia_corporativa").alias("corporativa"),

    F.sum("ia_codigo").alias("ia_codigo")

).show()
```

```python
ia_exec = gold_ia_exec.copy()

total_ia = ia_exec["count"].sum()

ia_exec["Percentual"] = (
    ia_exec["count"] / total_ia
) * 100


plt.figure(figsize=(12, 8))

ax = sns.barplot(
    data=ia_exec,
    x="categoria",
    y="Percentual",
    hue="categoria",
    palette="viridis",
    legend=False
)

plt.title(
    "Formas de Utilização de IA pelos Profissionais de Dados",
    fontsize=16,
    fontweight="bold",
    pad=15
)


plt.ylabel("Percentual dos Respondentes (%)")
plt.xlabel("")


for i, valor in enumerate(ia_exec["Percentual"]):
    ax.text(
        i,
        valor + 1,
        f"{valor:.1f}%",
        ha="center",
        fontsize=10,
        fontweight="bold"
    )

plt.tight_layout()
plt.show()
```

#### *Principais Achados*

- IA Gratuita é a principal porta de entrada.
- IA Corporativa apresenta forte adoção.
- Assistentes de código já fazem parte da rotina de muitos profissionais.
- Poucos participantes declararam não utilizar IA.

###**Uso de IA por Cargo**

### Pergunta de Negócio

Existe diferença na adoção de IA entre os diferentes perfis profissionais?

### Objetivo

Comparar a utilização de IA entre os principais cargos da área.

### Metodologia

Foram criados indicadores independentes para:

- IA Gratuita
- IA Corporativa
- IA para Código
- IA Paga Individual
- Não utiliza IA

As categorias não são mutuamente exclusivas.

```python
silver_ia_cargo = (
    silver_ia

    .withColumn(
        "cargo_limpo",
        F.when(
            F.col("cargo") == "Analista de Dados/Data Analyst",
            "Analista de Dados"
        )
        .when(
            F.col("cargo") == "Cientista de Dados/Data Scientist",
            "Cientista de Dados"
        )
        .when(
            F.col("cargo").isin(
                "Engenheiro de Dados/Data Engineer/Data Architect",
                "Engenheiro de Dados/Arquiteto de Dados/Data Engineer/Data Architect"
            ),
            "Engenheiro de Dados"
        )
        .when(
            F.col("cargo") == "Analista de BI/BI Analyst",
            "Analista de BI"
        )
        .when(
            F.col("cargo") == "Analista de Negócios/Business Analyst",
            "Business Analyst"
        )
        .when(
            F.col("cargo") == "Engenheiro de Machine Learning/ML Engineer/AI Engineer",
            "ML Engineer"
        )
        .when(
            F.col("cargo") == "Analytics Engineer",
            "Analytics Engineer"
        )
        .otherwise(F.col("cargo"))
    )
)
```

```python
cargos_principais = [
    "Analista de Dados",
    "Cientista de Dados",
    "Engenheiro de Dados",
    "Analista de BI",
    "Business Analyst",
    "Analytics Engineer",
    "ML Engineer"
]

base_ia_cargo = (
    silver_ia_cargo

    .where(
        F.col("cargo_limpo").isin(cargos_principais)
        &
        F.col("uso_ia").isNotNull()
    )
)
```

```python
base_ia_cargo.groupBy("cargo_limpo") \
    .count() \
    .orderBy(F.desc("count")) \
    .show(truncate=False)
```

```python
gold_cargo_ia = (
    base_ia_cargo

    .groupBy("cargo_limpo")

    .agg(
        F.count("*").alias("respondentes_ia"),

        F.sum("ia_gratuita").alias("ia_gratuita"),

        F.sum("ia_corporativa").alias("ia_corporativa"),

        F.sum("ia_codigo").alias("ia_codigo"),

        F.sum("ia_paga_individual").alias("ia_paga_individual"),

        F.sum("ia_nao_utiliza").alias("nao_utiliza")
    )

    .withColumn(
        "pct_ia_gratuita",
        F.round(
            F.col("ia_gratuita") / F.col("respondentes_ia") * 100,
            1
        )
    )

    .withColumn(
        "pct_ia_corporativa",
        F.round(
            F.col("ia_corporativa") / F.col("respondentes_ia") * 100,
            1
        )
    )

    .withColumn(
        "pct_ia_codigo",
        F.round(
            F.col("ia_codigo") / F.col("respondentes_ia") * 100,
            1
        )
    )

    .withColumn(
        "pct_ia_paga_individual",
        F.round(
            F.col("ia_paga_individual") / F.col("respondentes_ia") * 100,
            1
        )
    )

    .withColumn(
        "pct_nao_utiliza",
        F.round(
            F.col("nao_utiliza") / F.col("respondentes_ia") * 100,
            1
        )
    )

    .orderBy(F.desc("respondentes_ia"))
)
```

```python
gold_cargo_ia_pct = (
    gold_cargo_ia
    .select(
        "cargo_limpo",
        "respondentes_ia",

        F.round(
            F.col("ia_gratuita") /
            F.col("respondentes_ia") * 100,
            1
        ).alias("IA Gratuita"),

        F.round(
            F.col("ia_corporativa") /
            F.col("respondentes_ia") * 100,
            1
        ).alias("IA Corporativa"),

        F.round(
            F.col("ia_codigo") /
            F.col("respondentes_ia") * 100,
            1
        ).alias("IA para Código"),

        F.round(
            F.col("ia_paga_individual") /
            F.col("respondentes_ia") * 100,
            1
        ).alias("IA Paga Individual"),

        F.round(
            F.col("nao_utiliza") /
            F.col("respondentes_ia") * 100,
            1
        ).alias("Não utiliza IA")
    )
)

gold_cargo_ia_pct.show(truncate=False)
```

```python
cargo_ia_pd = gold_cargo_ia_pct.toPandas()

ordem_cargos = [
    "Analista de Dados",
    "Cientista de Dados",
    "Engenheiro de Dados",
    "Analista de BI",
    "Analytics Engineer",
    "Business Analyst",
    "ML Engineer"
]

cargo_ia_pd["cargo_limpo"] = pd.Categorical(
    cargo_ia_pd["cargo_limpo"],
    categories=ordem_cargos,
    ordered=True
)

cargo_ia_pd = (
    cargo_ia_pd
    .sort_values("cargo_limpo")
    .set_index("cargo_limpo")
)

matriz_ia = cargo_ia_pd[
    [
        "IA Gratuita",
        "IA Corporativa",
        "IA para Código",
        "IA Paga Individual",
        "Não utiliza IA"
    ]
]

plt.figure(figsize=(12, 6))

sns.heatmap(
    matriz_ia,
    annot=True,
    fmt=".1f",
    cmap="YlGnBu",
    linewidths=1.5,
    linecolor="white",
    annot_kws={
        "size":11,
        "weight":"bold"
    }
)

plt.title(
    "Modalidades de Uso de IA por Cargo",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Modalidade de Uso")
plt.ylabel("Cargo")

plt.xticks(rotation=20, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

### **Senioridade dos Profissionais de Dados**

### Pergunta de Negócio

Como está distribuída a senioridade dos profissionais de Dados no mercado brasileiro e como essa composição evoluiu entre 2023 e 2025?

### Objetivo

Identificar a composição do mercado por nível profissional e observar possíveis mudanças na distribuição de profissionais entre diferentes níveis de senioridade.

### Metodologia

A análise utiliza a variável **nível profissional** (`2.g_nivel`) presente nas bases consolidadas.

Os resultados são calculados como participação percentual dentro dos registros válidos de cada ano, permitindo comparar a composição da senioridade mesmo diante de diferenças no tamanho das amostras.



```python
col_nivel = F.col("`2.g_nivel`")

gold_senioridade = (
    df_total
    .select(
        "ano_pesquisa",
        col_nivel.alias("nivel")
    )
    .withColumn(
        "nivel",
        F.trim(F.col("nivel"))
    )
    .filter(
        F.col("nivel").isNotNull()
        & (F.col("nivel") != "")
        & (F.col("nivel") != "0")
        & (F.lower(F.col("nivel")) != "nan")
    )
)

# Quantidade de respostas por ano e nível
senioridade_ano = (
    gold_senioridade
    .groupBy(
        "ano_pesquisa",
        "nivel"
    )
    .count()
)

# Total válido por ano
total_senioridade_ano = (
    gold_senioridade
    .groupBy("ano_pesquisa")
    .count()
    .withColumnRenamed(
        "count",
        "total_ano"
    )
)

# Percentual por ano
gold_senioridade_pct = (
    senioridade_ano
    .join(
        total_senioridade_ano,
        "ano_pesquisa"
    )
    .withColumn(
        "percentual",
        F.round(
            F.col("count") / F.col("total_ano") * 100,
            2
        )
    )
    .orderBy(
        "ano_pesquisa",
        F.desc("percentual")
    )
)

gold_senioridade_pct.show(
    100,
    truncate=False
)
```

```python
# ==========================================
# GRÁFICO — DISTRIBUIÇÃO DE SENIORIDADE
# ==========================================

senioridade_pd = gold_senioridade_pct.toPandas()

# Tabela no formato:
# linhas = anos
# colunas = níveis
matriz_senioridade = (
    senioridade_pd
    .pivot(
        index="ano_pesquisa",
        columns="nivel",
        values="percentual"
    )
    .fillna(0)
)

# Ordenar anos
matriz_senioridade = matriz_senioridade.sort_index()

# Gráfico
ax = matriz_senioridade.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 7)
)

plt.title(
    "Composição de Senioridade dos Profissionais de Dados",
    fontsize=18,
    fontweight="bold",
    pad=18
)

plt.xlabel(
    "Ano da pesquisa",
    fontsize=12
)

plt.ylabel(
    "Participação dos profissionais (%)",
    fontsize=12
)

plt.ylim(0, 100)

plt.xticks(
    rotation=0
)

plt.legend(
    title="Nível profissional",
    bbox_to_anchor=(1.02, 1),
    loc="upper left",
    frameon=False
)

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.25
)

ax.grid(
    axis="x",
    visible=False
)

sns.despine()

plt.figtext(
    0.01,
    0.01,
    "Fonte: State of Data Brasil 2023–2025 | Base consolidada",
    fontsize=9,
    color="gray"
)

plt.tight_layout()

plt.show()
```

#### *Principais Achados*

A distribuição de senioridade permite identificar o grau de experiência predominante entre os profissionais participantes da pesquisa e possíveis mudanças na composição do mercado ao longo do período analisado.

### Conclusão

A análise de senioridade complementa a avaliação do mercado ao mostrar não apenas quais cargos estão presentes, mas também o nível de experiência dos profissionais que ocupam essas funções. Essa informação é relevante para estratégias de contratação, desenvolvimento e planejamento de carreira em uma instituição financeira.

### **Tecnologias Mais Adotadas**

### Pergunta de Negócio

Quais tecnologias apresentam maior adoção entre os profissionais de Dados?

### Objetivo

Identificar as tecnologias mais utilizadas pelos profissionais participantes da pesquisa e mapear o conjunto de ferramentas que apresenta maior presença no mercado brasileiro de Dados.

### Metodologia

As tecnologias são representadas na base por variáveis de seleção individual, nas quais cada coluna corresponde a uma tecnologia.

Foram consideradas as colunas pertencentes aos grupos de tecnologias identificados na estrutura da pesquisa. Para cada tecnologia, foi contabilizada a quantidade de respostas positivas e calculado o percentual de adoção entre os registros válidos.

Para evitar que tecnologias com poucas respostas dominem o ranking, são consideradas apenas variáveis com quantidade mínima de respostas válidas.


```python
# ==========================================
# FUNÇÃO PARA CALCULAR ADOÇÃO
# ==========================================

import re

def calcular_adocao(colunas, nome_grupo):

    resultados = []

    for c in colunas:

        # Remove o prefixo numérico da coluna
        tecnologia = re.sub(
            r"^4\.[a-z]\.\d+_",
            "",
            c
        ).strip()

        # Total de respostas válidas
        total = (
            df_total
            .filter(
                F.col(f"`{c}`").isNotNull()
            )
            .count()
        )

        # Respostas positivas (1)
        positivos = (
            df_total
            .filter(
                F.trim(F.col(f"`{c}`")) == "1"
            )
            .count()
        )

        if total > 0:

            percentual = round(
                positivos / total * 100,
                2
            )

            resultados.append(
                (
                    nome_grupo,
                    tecnologia,
                    total,
                    positivos,
                    percentual
                )
            )

    return resultados
```

```python
# ==========================================
# DEFINIÇÃO DAS CATEGORIAS DE TECNOLOGIA
# ==========================================

# LINGUAGENS
linguagens = [
    "SQL",
    "R",
    "Python",
    "C/C++/C#",
    ".NET",
    "Java",
    "Julia",
    "SAS/Stata",
    "Visual Basic/VBA",
    "Scala",
    "Matlab",
    "Rust",
    "PHP",
    "JavaScript"
]

colunas_linguagens = [
    c for c in df_total.columns
    if c.startswith("4.d.")
    and c.split("_", 1)[-1] in linguagens
]


# CLOUD
colunas_cloud = [
    c for c in df_total.columns
    if c.startswith("4.e.")
]


# BANCO DE DADOS / PLATAFORMAS
bancos = [
    "MySQL",
    "Oracle",
    "SQL SERVER",
    "Amazon Aurora ou RDS",
    "DynamoDB",
    "CoachDB",
    "Cassandra",
    "MongoDB",
    "MariaDB",
    "Datomic",
    "S3",
    "PostgreSQL",
    "ElasticSearch",
    "DB2",
    "Microsoft Access",
    "SQLite",
    "Sybase",
    "Firebase",
    "Vertica",
    "Redis",
    "Neo4J",
    "Google BigQuery",
    "Google Firestore",
    "Amazon Redshift",
    "Amazon Athena",
    "Snowflake",
    "Databricks",
    "HBase",
    "Presto",
    "Splunk",
    "SAP HANA",
    "Hive",
    "Firebird"
]

colunas_bancos = [
    c for c in df_total.columns
    if c.startswith("4.d.")
    and c.split("_", 1)[-1] in bancos
]


# BI / VISUALIZAÇÃO
colunas_bi = [
    c for c in df_total.columns
    if c.startswith("4.g.")
    and c.split("_", 1)[-1] not in bancos
]

print("Linguagens:", len(colunas_linguagens))
print("Cloud:", len(colunas_cloud))
print("Bancos/Plataformas:", len(colunas_bancos))
print("BI:", len(colunas_bi))
```

```python
# ==========================================
# RECALCULAR TECNOLOGIAS
# ==========================================

resultados = []

resultados += calcular_adocao(
    colunas_linguagens,
    "Linguagem"
)

resultados += calcular_adocao(
    colunas_cloud,
    "Cloud"
)

resultados += calcular_adocao(
    colunas_bancos,
    "Banco / Plataforma"
)

resultados += calcular_adocao(
    colunas_bi,
    "BI"
)

gold_tecnologias = pd.DataFrame(
    resultados,
    columns=[
        "grupo",
        "tecnologia",
        "respostas_validas",
        "adocao",
        "percentual_adocao"
    ]
)

gold_tecnologias = (
    gold_tecnologias
    .sort_values(
        ["grupo", "percentual_adocao"],
        ascending=[True, False]
    )
)

gold_tecnologias.head(40)
```

```python
# ==========================================
# GRÁFICO — TECNOLOGIAS MAIS UTILIZADAS
# ==========================================

import matplotlib.pyplot as plt
import textwrap

# ------------------------------------------
# CONFIGURAÇÃO VISUAL
# ------------------------------------------

cores = {
    "Linguagem": "#0B3C6D",          # Azul escuro
    "Cloud": "#2E8B57",              # Verde
    "Banco / Plataforma": "#E67E22", # Laranja
    "BI": "#6B4C9A"                  # Roxo
}

grupos = [
    "Linguagem",
    "Cloud",
    "Banco / Plataforma",
    "BI"
]

titulos = {
    "Linguagem": "Top 5 — Linguagens",
    "Cloud": "Top 5 — Cloud",
    "Banco / Plataforma": "Top 5 — Bancos / Plataformas",
    "BI": "Top 5 — Business Intelligence"
}

subtitulos = {
    "Linguagem": "Linguagens mais utilizadas",
    "Cloud": "Plataformas de cloud mais utilizadas",
    "Banco / Plataforma": "Bancos e plataformas mais utilizados",
    "BI": "Ferramentas de BI mais utilizadas"
}


# ------------------------------------------
# FIGURA
# ------------------------------------------

fig, axes = plt.subplots(
    2,
    2,
    figsize=(18, 12)
)

axes = axes.flatten()


# ------------------------------------------
# CADA CATEGORIA
# ------------------------------------------

for ax, grupo in zip(axes, grupos):

    # Seleciona grupo
    dados = (
        gold_tecnologias[
            gold_tecnologias["grupo"] == grupo
        ]
        .sort_values(
            "percentual_adocao",
            ascending=False
        )
        .head(5)
        .copy()
    )

    # Inverte para gráfico horizontal
    dados = dados.sort_values(
        "percentual_adocao",
        ascending=True
    )

    cor = cores[grupo]

    # --------------------------------------
    # QUEBRA DE NOMES LONGOS
    # --------------------------------------

    labels = [
        "\n".join(
            textwrap.wrap(
                str(nome),
                width=28
            )
        )
        for nome in dados["tecnologia"]
    ]

    # --------------------------------------
    # BARRAS
    # --------------------------------------

    barras = ax.barh(
        labels,
        dados["percentual_adocao"],
        color=cor,
        height=0.62,
        alpha=0.90
    )

    # --------------------------------------
    # TÍTULO
    # --------------------------------------

    ax.set_title(
        titulos[grupo],
        fontsize=16,
        fontweight="bold",
        loc="left",
        pad=20
    )

    # Subtítulo
    ax.text(
        0,
        1.02,
        subtitulos[grupo],
        transform=ax.transAxes,
        fontsize=10.5,
        color="#555555",
        ha="left",
        va="bottom"
    )

    # --------------------------------------
    # VALORES NAS BARRAS
    # --------------------------------------

    maior = dados["percentual_adocao"].max()

    for barra, valor in zip(
        barras,
        dados["percentual_adocao"]
    ):

        ax.text(
            valor + maior * 0.015,
            barra.get_y() + barra.get_height() / 2,
            f"{valor:.1f}%",
            va="center",
            ha="left",
            fontsize=11,
            fontweight="bold",
            color=cor
        )

    # --------------------------------------
    # EIXO X
    # --------------------------------------

    ax.set_xlabel(
        "Adoção (%)",
        fontsize=10
    )

    # Limite um pouco maior que a maior barra
    ax.set_xlim(
        0,
        maior * 1.18
    )

    # --------------------------------------
    # GRID
    # --------------------------------------

    ax.grid(
        axis="x",
        linestyle="--",
        linewidth=0.7,
        alpha=0.20
    )

    ax.set_axisbelow(True)

    # --------------------------------------
    # LIMPEZA VISUAL
    # --------------------------------------

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    ax.spines["left"].set_alpha(0.25)
    ax.spines["bottom"].set_alpha(0.25)

    ax.tick_params(
        axis="both",
        labelsize=10,
        length=0
    )


# ------------------------------------------
# TÍTULO PRINCIPAL
# ------------------------------------------

fig.suptitle(
    "Tecnologias mais utilizadas pelos profissionais de dados",
    fontsize=22,
    fontweight="bold",
    y=0.98
)

fig.text(
    0.5,
    0.945,
    "Top 5 por categoria — percentual de adoção entre as respostas válidas",
    ha="center",
    fontsize=12,
    color="#555555"
)


# ------------------------------------------
# RODAPÉ
# ------------------------------------------

fig.text(
    0.5,
    0.025,
    "Fonte: State of Data Brasil — edição mais recente disponível na base. "
    "Percentuais calculados sobre respostas válidas de cada pergunta.",
    ha="center",
    fontsize=9.5,
    color="#666666"
)


# ------------------------------------------
# AJUSTE FINAL
# ------------------------------------------

plt.subplots_adjust(
    top=0.88,
    bottom=0.09,
    left=0.08,
    right=0.97,
    hspace=0.38,
    wspace=0.28
)

plt.show()
```

#### *Principais Achados*

O ranking permite identificar o conjunto de tecnologias com maior presença entre os profissionais de Dados e evidenciar quais ferramentas possuem maior relevância no ecossistema analisado.

### Conclusão

A concentração de adoção em determinadas tecnologias evidencia um núcleo de competências recorrentes no mercado de Dados. Para uma instituição financeira, esse cenário pode orientar prioridades de capacitação, contratação e definição de tecnologias estratégicas.

# Arquitetura e Pipeline de Dados na AWS

A solução proposta utiliza uma arquitetura de processamento em camadas,
seguindo o conceito de Data Lake.

Os dados do State of Data Brasil 2023, 2024 e 2025 são inicialmente
armazenados no Amazon S3 na camada Bronze, preservando os dados em seu
formato original.

O AWS Glue é utilizado para catalogação e processamento dos dados,
utilizando Apache Spark/PySpark para as etapas de transformação e
padronização.

Na camada Silver são mantidas as informações relevantes para a análise,
após tratamento, padronização e harmonização das variáveis entre os
diferentes anos da pesquisa.

A camada Gold contém as tabelas analíticas utilizadas para responder às
questões de negócio, como estrutura do mercado, diversidade, região,
modelo de trabalho, adoção de IA, evolução dos cargos e remuneração.

O Amazon Athena permite realizar consultas SQL sobre os dados catalogados,
enquanto as ferramentas de visualização podem consumir as tabelas Gold
para construção dos dashboards e análises executivas.
